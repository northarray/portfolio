```
import pandas as pd

import numpy as np
```

Volatility Weighing Scheme

```
#| export
def volatility_matched(return_streams):
    vols = return_streams.std()
    inv_vols = 1 / vols
    return (inv_vols / inv_vols.sum()).to_dict()
```

Expects a pandas series with monthly returns data, each column representing an asset

```
dates = pd.date_range('2020-01', periods=24, freq='ME')

returns = pd.DataFrame({
    'Stocks': np.random.normal(0.01, 0.05, 24),   # higher vol
    'Bonds':  np.random.normal(0.004, 0.01, 24),  # lower vol
    'Gold':   np.random.normal(0.005, 0.03, 24),  # medium vol
}, index=dates)
```

```
returns.head()
```

```
weights = volatility_matched(returns)
weights
```
