## How to use

Check out core module for full usage. The main idea is that a portfolio consists of monthly return streams together with a weighing scheme. Here is a quick intro:

````python
rets, rf, cpi = sample_data_se()
````

````python
p = Portfolio('60/40', rets, {'bonds': .40, 'stocks': .60}, rf=rf, cpi=cpi)
p
````

````python
p.stats()
````

````python
p.real_w().head()
````

````python
p.real_w().plot(logy=True)
````
